# 1.인공지능

### ▶인공지능

**인공지능**(Artificial Intelligence: AI)이란 <font color="blue">기계가 학습과 추리를 통해서</font> 인간의 인지적인 기능을 흉내내어 문제를 해결하기 위하여 인간의 지능과 비슷한 작업을 수행하는 것을 의미한다. 예를들어, 글자나 말의 의미를 인식한다든지, 어떠한 내용을 학습한다든지, 얼굴 표정을 인지한다든지 즉, 인간만이 할 수 있다고 생각했던 기능을 컴퓨터가 수행하는 것이다.

## ▶인공지능 기술의 발전 과정



인공지능의 목표는 인간의 지능을 기계로 구현하는 것이다. 인간의 지능을 어떻게 구현하고 발현시킬 것인가에 대한 개념적 발전 흐름은 다음과 같다.

• 규칙 기반 AI (Rule-based AI)  

• 머신러닝 (Machine Learning)

• 딥러닝 (Deep Learning)

• 생성형 AI (Generative AI)
  



### 규칙 기반 AI (Rule-based AI)

- 기호와 규칙(Rule)을 사용하는 인공지능
- 1980년대 특히 성공적인 연구가 바로 전문가 시스템(expert system)이다. 전문가 시스템이란 **인간 전문가의 의사결정을 흉내내는 시스템**을 의미하는데, 규칙기반 시스템을 잘 활용한 경우라고 볼 수 있다.


### 머신러닝 (Machine Learning)

- 데이터를 통해 컴퓨터가 스스로 패턴을 학습하는 방식
- 데이터 증가와 함께 머신러닝 활용 확대
- 사람이 일일이 규칙을 정해주지 않아도, 데이터를 반복 학습하면서 규칙과 패턴을 찾음
- 스팸메일 필터, 추천 시스템, 주식 가격 예측

머신러닝과 딥러닝은 AI가 데이터를 학습하고 예측,분류하는 핵심 방법론이다.

### 딥러닝 (Deep Learning)

- 인공신경망 기반 학습 방법
- 인간 뇌의 신경세포(뉴런) 구조에서 영감을 받아 여러 층(layer)의 신경망을 구현 하고 데이터를 학습하는 인공신경망(Artificial Neural Network) 사용
- 머신러닝은 사람이 특징을 뽑아야 하지만(사람이 특징을 설계), 딥러닝은 모델이 스스로 특징을 학습
- 이미지 인식, 음성 인식 등에서 높은 성능
  - 컴퓨터 비전(CV, Computer Vision): 이미지와 영상을 인식하고 이해, 얼굴 인식, 자율주행차 카메라, 스마트 CCTV

  - 자연어 처리(NLP, Natural Language Processing): 인간 언어를 이해하고 생성, 번역기, 챗봇, 외국어, 의료

  - 음성 인식(Speech Recognition): 사람의 음성을 문자로 바꾸거나 이해, Siri, 구글 어시스턴트, 가전제품
- 대규모 데이터와 컴퓨팅 자원의 발전  

<div align="center"><img src="https://haesunbyun.github.io/common/images/neuron.png" ></div>

- 인간신경망의 형태
- Dendrite(가지돌기 또는 수상돌기): 신경세포(뉴런)에서 뻗어나온 나뭇가지 모양의 짧은 돌기
- Axon(축삭돌기): 신경 세포의 세포체에서 길게 뻗어 나온 가지. 활동전위를 전달하는 역할
- 축삭은 일반적으로 뉴런 하나당 하나씩만 존재하며, 길이는 수 밀리미터에서 수 센티미터에 이르기도 함.
- 축삭의 끝은 다른 뉴런의 수상돌기(dendrite)와 연결되어 있는데, 이를 시냅스(synapse)라 함



<div align="center"><img src="https://haesunbyun.github.io/common/images/aineuron.png" ></div>

- 인간신경망의 뉴런을 인공적으로 만들어서 사람이 학습하듯이 기계가 학습하게 만들자는 아이디어가 인공 신경망으로 이런 모양

- 인공신경망은는 다음과 같은 구조를 띤다.

<div align="center"><img src="https://haesunbyun.github.io/common/images/aineuron2.jpg" ></div>

- Input layer: 외부에서 제공되는 데이터를 받아들이는 역할.
주로, 데이터의 특징(feature)을 나타내는 노드들로 이루어짐. 예를 들어, 이미지를 처리하는 경우, 각 픽셀 값이 입력층의 노드로 들어갈 수 있음

- Hidden layer: 입력층에서 받은 정보를 처리하고 추상화하는 역할. 모델이 데이터의 복잡한 관계를 학습하고 표현할 수 있도록 함

- Output layer: 최종적으로 모델이 내놓는 결과

### 규칙 기반 AI vs. 머신러닝 vs. 딥러닝

→ 규칙 기반 AI : 사람이 규칙을 만든다.

예)
- 어떤 단어가 나오면 스팸인지
- 어떤 조건을 만족하면 스팸인지

→ 머신러닝 : 사람이 특징을 설계한다.

예)
- 제목 길이
- 특정 단어 등장 횟수
- 발신자 이메일 도메인
- 링크 개수
이 특징들을 모델에 넣으면, 모델이 스스로 스팸 판단 규칙을 학습

→ 딥러닝에서는 사람이 특징을 설계하지 않아도
모델이 데이터에서 특징을 자동으로 학습

# 2.생성형 AI



## ▶생성형 AI의 정의와 특징



- 생성형 AI의 정의

  - 기존 데이터를 학습한 후 새로운 콘텐츠를 만들어내는 인공지능
  - 단순히 기존 데이터를 분류하거나 예측하는 것이 아니라, 완전히 새로운 텍스트·이미지·음악 등을 생성
    - ChatGPT는 수많은 문서를 학습한 후 새로운 문장을 생성
    - 텍스트 설명만으로 새로운 그림을 창작

- 배경

  • 대규모 데이터의 축적

  • 컴퓨팅 성능의 향상

  • 대규모 언어모델(LLM)의 등장


- 생성형 AI의 특징

  - 새로운 콘텐츠 생성
  - 자연어 기반 상호작용
  - 다양한 분야에서 활용 가능
  - 단순한 검색 결과와 달리, 원본에 없는 새로운 결과물이라는 점에서 큰 차이
  - 따라서 생성형 AI는 인간의 창의적 활동을 보조하거나 확장하는 기술로 주목



- 전통적 AI와 차이점
  - 전통적 AI는 데이터를 분류하거나 패턴을 인식하는 데 집중
  - 생성형 AI는 데이터를 학습하여 새로 창조하는 데 집중
    - 사진 속에 고양이가 있는지 없는지를 판별하는 것이 전통적 AI의 역할
    - 생성형 AI는 사진 속 고양이를 인식하는 것이 아니라, 새로운 고양이 이미지를 창작
  - AI가 단순히 답을 알려주는 것을 넘어, 레포트 초안을 작성하거나 새로운 아이디어를 제안


## ▶생성형 AI의 핵심기술




### 대규모 언어모델 (LLM, Large Language Model)
- LLM은 수십억 개 이상의 단어와 문장을 학습하여 새로운 문장을 만들어내는 모델
  - GPT, Claude, LLaMA

- 이들은 문장의 “다음 단어”를 예측하는 방식으로 문장을 완성
- 단순 규칙 기반 문장 생성하는 것이 아니라, 방대한 데이터를 기반으로 학습 패턴으로 생성함
- 사용자가 질문을 하면 자연스럽고 논리적인 대답

### 확산 모델 (Diffusion Model)
- 이미지와 영상을 만들어내는 데 많이 사용
- 원리는 노이즈(잡음)로 가득 찬 이미지를 점차적으로 선명하게 변환
  - DALL·E, Stable Diffusion, Midjourney


- 기존에 존재하지 않던 장면을 새롭게 만들어낼 수 있다는 것이 가장 큰 장점
- 광고, 디자인, 예술 창작 등 다양한 분야에서 각광받고 있음
