## 😄 ChatGPT 활용 BMI 계산 프로그램
😄신장(cm)과 체중(kg)을 입력 받아 아래 BMI(신체질량지수) 공식을 이용하여 BMI를 계산하는 프로그램을 작성하시오. BMI가 25이상이면 비만입니다. * 단위에 주의하세요.

---

```
weight(kg)
height(m)

```
$ BMI = weight / height^2 $


<font color='blue'>`test case-1`</font>

Input Sample
```
160
65.2
```
Output Sample
```
신장(cm):
체중(kg):
BMI는 25.47
비만입니다. 운동으로 감량이 필요합니다.
```
<font color='blue'>`test case-2`</font>

Input Sample
```
170
55.3
```
Output Sample
```
신장(cm):
체중(kg):
BMI는 19.13
정상입니다. 꾸준한 운동으로 유지시켜주세요.
```
