## 문자열 처리함수

|메서드|설명|예제|결과|
|:----------:|:----------|:----------|:----------|
|capitalize|첫 문자를 대문자로 변환|'python'.capitalize()|'Python'|
|count|특정 문자의 개수를 반환|'banana'.count('a')|3|
|startswith|문자열이 특정 문자로 시작하면 True 반환|'hello'.startswith('he')|True|
|endswith|문자열이 특정 문자로 끝나면 True 반환|'hello'.endswith('lo')|True|
|find|찾을 문자열의 첫 번째 인덱스를 반환, 실패 시 -1 반환|'hello'.find('e')|1|
|index|find 메서드와 같으나 실패 시 ValueError 반환|'hello'.index('e')|1|
|isalpha|모든 문자가 알파벳이면 True 반환|'hello'.isalpha()|True|
|isdecimal|모든 문자가 숫자이면 True 반환|'1234'.isdecimal()|True|
|isalnum|모든 문자가 알파벳이거나 숫자이면 True 반환|'abc123'.isalnum()|True|
|lower|모든 문자를 소문자로 변환|'HELLO'.lower()|'hello'|
|upper|모든 문자를 대문자로 변환|'hello'.upper()|'HELLO'|
|replace|문자열의 문자를 다른 문자로 교체|'banana'.replace('a', 'o')|'bonono'|
|strip|문자열의 맨 앞과 맨 뒤에 있는 빈 칸을 제거|'  hello  '.strip()|'hello'|
|split|구분자를 지정하여 문자열을 나누고 리스트로 반환|'a,b,c'.split(',')|['a', 'b', 'c']|
|partition|문자열을 첫번째 파라미터로 분리하여 튜플로 반환|'hello world'.partition(' ')|('hello', ' ', 'world')|
|center|지정한 너비로 문자열을 늘이고 가운데 정렬|'hello'.center(10)|'  hello   '|
|zfill|문자열의 빈 칸을 '0'으로 채움|'42'.zfill(5)|'00042'|




````python
'python'.capitalize()     # 첫 문자를 대문자로 변환
'banana'.count('a')       # 'a'의 개수를 반환
'hello'.startswith('he')  # 문자열이 'he'로 시작하면 True 반환
'hello'.endswith('lo')    # 문자열이 'lo'로 끝나면 True 반환
'hello'.find('e')         # 'e'의 첫 번째 인덱스를 반환, 실패 시 -1 반환
'hello'.index('e')        # 'e'의 첫 번째 인덱스를 반환, 실패 시 ValueError 반환
'hello'.isalpha()         # 모든 문자가 알파벳이면 True 반환
'1234'.isdecimal()        # 모든 문자가 숫자이면 True 반환
'abc123'.isalnum()        # 모든 문자가 알파벳이거나 숫자이면 True 반환
'HELLO'.lower()           # 모든 문자를 소문자로 변환
'hello'.upper()           # 모든 문자를 대문자로 변환
'banana'.replace('a', 'o') # 'a'를 'o'로 교체
'  hello  '.strip()       # 문자열의 맨 앞과 맨 뒤에 있는 빈 칸을 제거
'a,b,c'.split(',')        # ','를 기준으로 문자열을 나누고 리스트로 반환
'hello world'.partition(' ') # 첫 번째 공백을 기준으로 분리하여 튜플로 반환
'hello'.center(10)        # 지정한 너비로 문자열을 늘이고 가운데 정렬
'42'.zfill(5)             # 문자열의 빈 칸을 '0'으로 채움
````

## 정규식 패턴 실습

https://regex101.com/

|문법|설명|예제|결과|
|:----------:|:----------|:----------|:----------|
|. |특정 문자 앞에 문자 1개가 있는 패턴을 찾음|'.a'|'ba'에서 'a'를 찾음|
|^ |문자열의 처음부터 일치하는 패턴을 찾음|'^I like'|'I like pizza'에서 'I like'를 찾음|
|$ |문자열의 끝 부분부터 일치하는 패턴을 찾음|'on\$'|'I am on'에서 'on'을 찾음|
|* |문자 뒤에 0개 이상의 특정 문자가 오는 패턴을 찾음|'n\d*'|'n123'에서 'n123'을 찾음|
|+ |문자 뒤에 1개 이상의 특정 문자가 오는 패턴을 찾음|'n\d+'|'n123'에서 'n123'을 찾음|
|? |앞의 문자가 있거나 없는 패턴을 찾음|'apple?'|'apple'과 'appl' 둘 다 찾음|
|{m} |문자 뒤에 특정 개수만큼 오는 패턴을 찾음|'n\d{2}'|'n12'에서 'n12'를 찾음|
|{m,n} |문자 뒤에 특정 개수 범위 내에서 오는 패턴을 찾음|'n\d{2,4}'|'n1234'에서 'n1234'를 찾음|
|\ |특수 문자를 일반 문자로 검색할 때 사용|'\*', '\?', '\+'|'*', '?', '+'를 그대로 찾음|
|[ ] |대괄호 안의 문자 중 하나를 포함하는 패턴을 찾음|'[cfh]all'|'call', 'fall', 'hall'에서 찾음|
| \| |'apple 또는 application' 연산자로 둘 중 하나의 패턴을 찾음| 'apple \| application' |'apple' 또는 'application'을 찾음|
|( ) |그룹 패턴을 지정하여 찾음|'(\d+)-(\d+)-(\d+)'|'123-456-789'에서 '123', '456', '789'를 그룹으로 찾음|


|특수문자|설명|예제|결과|
|:----------:|:----------|:----------|:----------|
|\d |숫자 1개를 의미|'\d'|'123'에서 '1', '2', '3'을 각각 찾음|
|\D |숫자 이외의 문자 1개를 의미|'\D'|'abc123'에서 'a', 'b', 'c'를 각각 찾음|
|\s |공백 문자 1개를 의미|'\s'|'a b'에서 공백을 찾음|
|\S |공백 문자 이외의 문자 1개를 의미|'\S'|'a b'에서 'a'와 'b'를 각각 찾음|
|\w |글자 또는 숫자 1개를 의미|'\w'|'a1!'에서 'a', '1'을 찾음|
|\W |글자 또는 숫자 이외의 문자 1개를 의미|'\W'|'a1!'에서 '!'을 찾음|


````python
import re

s = 'aaa@xxx.com, bbb@yyy.com'
m = re.match(r'([a-z]+)@([a-z]+)\.com', s)
print(m)
````

````python
if m:
  print('match')
else:
  print('not match')
````

````python
m.group()
````

````python
m.group(0)
m.group(1)
m.group(2)
````

### compile 메서드
- 정규식 객체를 리턴해 줌.
- 패턴을 반복해서 사용할 경우에 유용


````python
s = 'aaa@xxx.com, bbb@yyy.com, ccc@zzz.net'
p = re.compile(r'([a-z]+)@([a-z]+)\.com')
print(type(p))
````

````python
m = p.match(s)
print(m)
````

````python
m.group()
````

````python
result = p.sub('new-address', s)
print(result)
````
