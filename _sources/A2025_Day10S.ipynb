### 데이터 준비하기

- MNIST 데이터셋을 읽어온다.

````python
mnist = tf.keras.datasets.mnist

(x_train, y_train), (x_test, y_test) = mnist.load_data()
````

- x_train은 이미지데이터
- 학습용 데이터에 60,000개의 이미지
- 각 이미지는 28x28 개의 픽셀로 구성
- 이미지 데이터의 예)
  - 배경은 0이고 그레이스케일로 1에서 255까지의 값이 있는 부분이 글씨임

````python
x_train.shape, y_train.shape
````

<div align="center"><img src="https://haesunbyun.github.io/common/images/deep2.jpg" ></div>

````python
print(x_train[0])
````

- 위 데이터를 이미지로 확인해보면 다음과 같다.

````python
import matplotlib.pyplot as plt

plt.imshow(x_train[0], cmap=plt.cm.binary)
plt.show()
````

- y_train에는 답이 들어있다. 확인해보자.


````python
print(y_train[0])
````

- 신경망을 훈련하기 전에 이 데이터를 각 픽셀이 0에서 1까지의 값을 갖도록 바꾸어 준다.

````python
x_train, x_test = x_train / 255.0, x_test / 255.0
````

### 신경망 만들기

- 훈련용 데이터를 이용해서 숫자를 읽을 수 있도록 신경망을 만들고 학습시켜보자.
- 사용할 신경망은 keras 라이브러리의 모델 중에 sequential 모델

<div align="center"><img src="https://haesunbyun.github.io/common/images/MNIST_neuralnet_image.png" width="500" ></div>


- Keras의 sequential모델을 이용해서 신경망을 만든다.

````python
model = tf.keras.models.Sequential([
  tf.keras.layers.Flatten(input_shape=(28, 28)),
  tf.keras.layers.Dense(100, activation='relu'),
  tf.keras.layers.Dropout(0.2),
  tf.keras.layers.Dense(10, activation='softmax')
])

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])
````

### 신경망 학습시키기


````python
model.fit(x_train, y_train, epochs=5)
````

- 학습이 끝나면 새로운 이미지를 읽을 수 있는지 테스트해 보자.

````python
predictions = model.predict(x_test)
````

````python
plt.imshow(x_test[0], cmap=plt.cm.binary)
plt.show()
````

- 각 숫자에 대한 확률을 나타내는 값

````python
print (predictions[0])
````

- 이 중 확률이 가장 높다고 예측된 값이 신경망이 낸 답: 7

````python
import numpy as np
print(np.argmax(predictions[0]))
````

- 학습된 숫자인식 모델은 이 데이터셋에서 약 97%의 정확도를 보여줌

````python
model.evaluate(x_test, y_test)
````

````python
img = cv2.imread('night.jpg', cv2.IMREAD_COLOR)
plt.axis('off')
plt.imshow(img)
plt.show()
````

- 이미지의 색 값을 히스토그램으로 나타내보자.
- numpy의 ravel()함수는 1차원 배열로 변환시켜주는 함수이다.
- 이미지는 3차원 배열로 저장되어 있으므로 이를 1차원 배열로 변환한 후, 히스토그램을 그린다.
- 색이 얼마나 한쪽으로 쏠려있는지 확인할 수 있다.
- y축 좌표위의 1e7은 1*10^7이다.

````python
plt.figure(figsize=(7,5))
plt.hist(img.ravel(), bins=32)
plt.xlim([0,256])
plt.show()
````

- 한쪽으로 쏠려있는 이미지를 평평하게 펴주자.
- split() 함수를 이용하여 컬러별로 분리한후 b, g, r로 저장한다.
- equalizeHist()를 통해 b,g,r의 색을 평평하게 편다.
- 그런 후에 merge()함수를 통해 다시 합친다.

````python
b, g, r = cv2.split(img)
b1 = cv2.equalizeHist(b)
g1 = cv2.equalizeHist(g)
r1 = cv2.equalizeHist(r)
img_eq = cv2.merge((b1,g1,r1)) #변형된 각각 채널을 합침

plt.axis('off')
plt.imshow(img_eq)
plt.show()
#cv2_imshow(img_eq)
````

- 평평하게 펴진 히스토그램을 확인해보자.

````python
plt.figure(figsize=(7,5))
plt.hist(img_eq.ravel(), bins=32)
plt.xlim([0,256])
plt.show()
````

````python
img = cv2.imread('night.jpg', cv2.IMREAD_COLOR)

#HSV 색상(Hue), 채도(Saturation), 명도(Value) 채널분리 후 value 채널만 평평히 편다.
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
h,s,v = cv2.split(hsv)
v1 = cv2.equalizeHist(v)
hsv1 = cv2.merge((h,s,v1)) #변형된 Value 채널을 합침
img_hsv_eq = cv2.cvtColor(hsv1, cv2.COLOR_HSV2BGR)

plt.axis('off')
plt.imshow(img_hsv_eq)
plt.show()
````

#### 동영상에서 보행자 검출

- cv2.HOGDescriptor()에서 HOG는 Histogram of Gradient로 HOGDescriptor()는 객체를 검출하기 위해 만들어진 함수이다.
- setSVMDetector()에서 SVM은 Supprt Vector Machine의 약자로, 여백을 최대화하여 초평면을 찾는 지도학습 알고리즘이다.
- setSVMDetector()는 SVM알고리즘으로 이동하는 물체를 추적하기 위해 미리 학습된 모델이다.
- setSVMDetector() 전달인자로 HOGDescriptor_getDefaultPeopleDetector()를 인자로 넣어 사람을 검출하도록 한다.

````python
hog = cv2.HOGDescriptor()
hog.setSVMDetector(cv2.HOGDescriptor_getDefaultPeopleDetector())
````

- cv2모듈의 VideoCapture()로 동영상을 불러온다.
- 반복문을 통해 동영상에서 프레임을 읽어온다. 프레임이 제대로 읽혔다면 즉 ret==True이라면, 그 프레임에서 보행자를 검출한다.
- 검출결과를 화면에 표시하기 위해 검출좌표를 사각형으로 표시한다.

- [동영상 다운로드](https://github.com/HaesunByun/common/blob/d5ef525f60459671d4fd6c3f8676394dd8587c01/data/walking.avi?raw=True)

````python
from IPython.display import clear_output

cap = cv2.VideoCapture('walking.avi')

while True:
    ret, frame = cap.read()
    if not ret:
        break
    detected, _ = hog.detectMultiScale(frame) # 매 프레임마다 보행자 검출

    # 검출 결과 화면 표시
    # 사각형 시작좌표 x,y, 종료좌표 w,h, 사각형색깔(127,0,0) , 사각형 두께 2
    for (x, y, w, h) in detected:
        cv2.rectangle(frame, (x, y, w, h), (127, 0, 0), 2)

    clear_output(wait=True)
    cv2_imshow(frame)
    # if cv2.waitKey(10) == 27:
    #     break
````

````python
import speech_recognition as sr
r = sr.Recognizer()
````

- 음성파일을 다운로드 한다. [음성 파일 다운로드](https://github.com/HaesunByun/common/blob/289fabaa9f56a2ecc51f4b7f6d32493b7a4c0ef0/data/harvard.wav?raw=True)
- 파일을 open()하면 반드시 close()해 줘야 하는데, with로 파일을 열면 with를 벗어날 때 자동으로 파일이 close 되어 별도로 close()를 해줄 필요가 없다.
```
with open(파일 경로, 모드) as 파일 객체:
    처리 코드
```
- 음성인식 객체인 r의 record()를 통해 오디오 데이타를 얻는다.



````python
hello = sr.AudioFile('harvard.wav')

with hello as source:
    audio = r.record(source)
````

- recognize_google()은 google의 web speech API이다. 기본 API키가 있어서 별도의 API인증키없이 사용 가능하다.
- recognize_google()를 통해 인식된 음성 결과를 텍스트로 반환한다. 옵션으로 언어를 설정해준다.

````python
r.recognize_google(audio, language='eng')
````

- record() 옵션으로 duration을 주면 해당 초 만큼만 끊어서 텍스트로 읽어온다.

````python
harvard = sr.AudioFile('harvard.wav')
with harvard as source:
    audio1 = r.record(source, duration=4)
    audio2 = r.record(source, duration=4)
````

````python
r.recognize_google(audio1)
````

````python
r.recognize_google(audio2)
````

- record() 옵션으로 offset을 주면 해당 초 후부터 읽어온다.

````python
with harvard as source:
    audio = r.record(source, offset=4, duration=3)
````

````python
r.recognize_google(audio)
````

## 자연어처리를 위한 NLTK
- NLTK(Natural Language Toolkit) 자연어 처리 패키지
  - 자연어 처리 및 문서 분석용 파이썬 패키지

- NLTK 패키지가 제공하는 주요 기능
  - 말뭉치: 자연어 분석 작업을 위해 만든 샘플 문서 집합
  - 토큰 생성: 문자열을 토큰으로 나누는 작업
  - 형태소 분석: 단어로부터 어근, 접두사, 접미사, 품사 등 다양한 언어적 속성을 파악하고 이를 이용하여 형태소를 찾아내거나 처리하는 작업
  - 품사 태깅: 낱말을 문법적인 기능이나 형태, 뜻에 따라 구분

````python
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.tag import pos_tag
from nltk.chunk import ne_chunk

# NLTK 데이터 다운로드 (최초 실행 시 필요)
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')
nltk.download('maxent_ne_chunker')
nltk.download('words')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('maxent_ne_chunker_tab')

# 입력 텍스트
text = "Apple is looking at buying U.K. startup for $1 billion. Microsoft was founded by Bill Gates and Paul Allen."

# 문장 토큰화
sentences = sent_tokenize(text)
print("문장 토큰화:")
print(sentences)

# 단어 토큰화 및 품사 태깅
for sentence in sentences:
    words = word_tokenize(sentence)
    tagged_words = pos_tag(words)
    print("\n단어 토큰화 및 품사 태깅:")
    print(tagged_words)

    # 명명된 개체 인식
    named_entities = ne_chunk(tagged_words)
    print("\n명명된 개체 인식:")
    print(named_entities)
````

## 자연어 생성 샘플 코드
- 'gpt2':인터넷에서 수집한 대규모 텍스트 데이터셋(예: 웹페이지, 뉴스 기사, 책 등)을 사용하여 학습된 모델
- 학습의 목표는 주어진 텍스트 시퀀스에서 다음 단어를 예측

````python
from transformers import GPT2LMHeadModel, GPT2Tokenizer

# 모델과 토크나이저를 로드합니다.
model_name = 'gpt2'  # 'gpt2-medium', 'gpt2-large', 'gpt2-xl' 등으로 변경 가능
tokenizer = GPT2Tokenizer.from_pretrained(model_name)
model = GPT2LMHeadModel.from_pretrained(model_name)

# 입력 텍스트 설정
input_text = "Once upon a time"

# 입력 텍스트를 토큰화합니다.
input_ids = tokenizer.encode(input_text, return_tensors='pt')

# 텍스트를 생성합니다.
output = model.generate(input_ids, max_length=100, num_return_sequences=1)

# 생성된 텍스트를 디코딩합니다.
generated_text = tokenizer.decode(output[0], skip_special_tokens=True)

print("생성된 텍스트:")
print(generated_text)
````
