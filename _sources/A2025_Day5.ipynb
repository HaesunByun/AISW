## 배열과 리스트의 비교
- 얼핏 비슷해 보이지만 완전히 기능이 다른
파이썬의 리스트(list)와 numpy의 배열(array)의 차이를 간단히 비교해보자.

- 리스트와 배열의 생성에서,
  - 리스트는 정수, 실수, 문자열, 리스트 등 다양한 자료형을 담을 수 있다.
  - 배열은 정수형 또는 실수형만 담을 수 있는데 하나의 배열에는 하나의 자료형만 담을 수 있다.


````python
its_list = ['바나나', '딸기', 100, 5.5,['BTS', '블랙핑크']]
its_array = np.array([10,20,30])
````

  - 리스트와 배열을 출력했을 때,
    - 리스트는 대괄호안에 콤마(,)로 아이템을 구분한다.
    - 배열은 대괄호안에 구분기호 없이 공백으로만 원소를 구분한다.

````python
print(its_list)
print('========')
print(its_array)
````

- 리스트와 배열의 타입은,
  - 리스트는 list 타입이다.
  - 배열은 numpy.ndarray 타입이다.

````python
print(type(its_list))
print('========')
print(type(its_array))
````

- 리스트와 배열에 곱셈연산을 해보았을 때,
  - 리스트에서 곱셈은 리스트를 이어 붙이는 concatenate이다. 즉, 각 원소에 곱셈을 하지 않는다.
  - 배열은 각 원소에 벡터연산을 한다.


````python
print(its_list*2)
print('========')
print(its_array*2)
````

- 리스트와 배열의 메서드
  - 리스트와 배열은 자료형이 다르므로 지원하는 메서드의 종류도 다르다.
  - 예를 들어, 리스트에서는 mean, argmax, round 등의 메소드를 지원하지 않는다.
    - maen: 평균
    - argmax: 최대값 인덱스
    - round: 숫자를 지정한 자리수로 반올림
    - ...
  - 반면, 배열에서는 append, remove, extend 등 리스트 메소드를 지원하지 않는다.
    - append: 원소 추가
    - remove: 원소 삭제
    - extend: 리스트 확장

- 리스트와 배열의 연산속도
  - 리스트와 배열의 연산속도를 확인해보자.
  - 억개의 숫자에 대해서 2를 곱했다.
  - 결과를 보면 배열(array)가 10배 이상 빠르다.
  - 배열은 동일한 자료형만 저장하므로 데이터 저장에 필요한 저장공간이 일정하며 연속적이다. 따라서 메모리 주소 측면에서 인덱스를 통해 빠르게 데이터에 접근할 수 있다.
  - 반면, 리스트는 다양한 자료형을 저장할 수 있어서 필요한 저장공간이 일정하지 않으며, 메모리 활동공간도 연속적이지 않다. 인덱스를 갖지만 이는 몇 번째 데이터인가 정도를 의미한다.

```python
import time
a = list(range(10 ** 8))

start = time.time()
[i *2 for i in a]
end = time.time()
print(f'리스트 실행시간: {end - start}')

a = np.array(range(10 ** 8))
start = time.time()
a * 2
end = time.time()
print(f'배열 실행시간: {end - start}')
```



## 인덱싱
- 1차원 배열을 인덱싱하는 것은 리스트와 동일하다.


````python
ar=np.arange(12)
print(ar, '\n\n===결과===')
print(ar[5])
````

- 2차원 배열을 인덱싱하는 방법은 대괄호를 연속하여 쓰고 인덱스를 각각 넣거나 하나의 대괄호에 콤마로 구분하여 인덱스를 넣어준다.
    - 배열명[행인덱스][열인덱스]
    - 배열명[행인덱스,열인덱스]  

- 대괄호를 연속하여 쓰면 가장 앞 대괄호부터 인덱싱 또는 슬라이싱 해서 나온 결과를 가지고 다음 대괄호의 인덱싱 또는 슬라이싱을 차례대로 한다.
즉, 앞의 대괄호에서 나온 결과가 작업범위의 대상이 된다.
- 예를 들어, 아래와 같다.

```python
 ar2 = ar[1]
 ar3 = ar2[2]
 ar3 == ar[1][2]
```
- 대괄호를 하나 쓰는 것은 작업범위가 배열 전체를 대상으로 한다.
- 인덱싱에서는 대괄호를 연속으로 써는 방법이나 하나의 대괄호에 콤마로 구분하여 쓰는 방법이나 인덱싱의 결과는 똑같다. 그러나 슬라이싱에서는 그 결과가 달라진다.

````python
ar=np.arange(12).reshape(4,3)
print(ar, '\n\n===결과===')
print(ar[1][2])
print(ar[1,2])
````

````python
ar=np.arange(12).reshape(4,3)
ar2 = ar[1]
ar3 = ar2[2]
ar3 == ar[1][2]
````

- 3차원의 경우에도 2차원에서와 마찬가지로 면, 행, 열 순서로 사용하면 된다.

````python
ar=np.arange(12).reshape(2,2,3)
print(ar, '\n\n===결과===')
print(ar[1][1][2])
print(ar[1,1,2])
````

## 슬라이싱
- 슬라이싱의 경우에는 대괄호를 연속으로 사용하는 것과 하나의 대괄호에 콤마로 나열하는 것의 결과가 다를 수도 있다.

````python
ar=np.arange(12).reshape(3,4)
print(ar)
print('\n\n===연속된 대괄호 결과===')
print(ar[1:][1:])
print('\n\n===하나의 대괄호 결과===')
print(ar[1:, 1:])
````

- 위의 코드에서 ar[1:]의 결과는 아래와 같다.

```
 [[ 4  5  6  7]
 [ 8  9 10 11]]
```
- 이 결과를 변수 a에 저장했다면 ar[1:][1:]는 a[1:]와 같다. 즉, a의 1행부터 끝까지 이므로 [ 8  9 10 11]을 슬라이싱 한다.

- ar[1:, 1:]는 ar 전체를 대상으로 작업하므로 행은 1행부터 끝까지, 그 범위에서 열이 1열부터 끝까지를 슬라이싱한다.

where(): 조건에 만족하는 인덱스를 반환하거나, 조건에 따라 값을 다르게 적용시킨다.
```
np.where(조건, True일때 적용값, False일때 적용값)
```

````python
a = np.arange(10)
print(a)
print('\n\n===결과===')
print(np.where(a%2==0)) # index 반환
print(np.where(a%2==0, "짝수", "홀수"))
````

### sum(), mean(), median()
- np.sum(배열명): 합을 반환한다.

## Deep Copy vs. Shallow Copy
배열의 데이터를 복사해서 복사된 데이터를 가지고 작업을 해야 할 때가 있다.
배열을 복사하는 방법에는 두 가지가 있다.

- Shallow Copy : 원조변수와 복사변수가 같은 메모리를 참조
- Deep Copy : 복사변수는 새로이 메모리를 할당하여 배열을 생성

- `cp_ar = ar`와 같은 복사를 shallow copy라고 한다.
- shallow 복사의 문제는 복사 배열의 데이터를 작업을 하면 원조 배열의 데이터도 똑같이 작업된다는 것이다.
- 그 이유는 복사변수 cp_ar과 원조변수 ar이 같은 메모리 주소를 참조하기 때문이다.

````python
import numpy as np
ar = np.arange(1,10).reshape(3,3)
print(ar)

cp_ar = ar  #shallow copy
print(cp_ar)

cp_ar[0][2]=1000
print(cp_ar,'\n\n', ar) # cp_ar의 값을 바꿨는데 ar도 바뀌어 있음
````


- 이를 방지하기 위해서는 cp_ar = ar.copy()와 같이 Deep Copy를 해야 한다.
- Deep copy는 복사변수의 값을 작업(수정, 삭제, 추가 등)해도 원본변수의 값이 작업되지 않는다.

````python
ar = np.arange(1,10).reshape(3,3)
print(ar)

cp_ar = ar.copy() #Deep copy
print(cp_ar)

cp_ar[0][2]=1000
print(cp_ar,'\n\n', ar) # 복사된 곳에서 값을 수정해도 원본에 영향없음
````
