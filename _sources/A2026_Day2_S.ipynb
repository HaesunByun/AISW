# 2.생성형 AI: 텍스트

## ChatGPT란

- ChatGPT는 미국의 AI 연구재단 OpenAI가 2022년 11월 공개한 초거대 언어모델 GPT-3.5 기반 대화형 인공지능 챗봇
- ChatGPT는 'Generative Pre-trained Transformer'의 약자
- Generative(생성형), Pre-trained(미리 학습된) Transformer(변환기)


- 현재 ChatGPT 홈페이지(https://chatgpt.com/) 에서 누구나 가입 후 사용 가능.

- 사용자가 채팅 하듯 질문을 입력하면 ChatGPT는 학습한 데이터를 기반으로 '사람처럼' 문장을 만들어 답을 해줌
- 마치 진짜 사람과 대화하는 것처럼 자연스럽게 질문과 답변을 주고받을 수 있음
- 단순 정보 짜깁기뿐 아니라 에세이・소설・시 등 다양한 창작물, 심지어 프로그래밍 코드까지 생성한다는 점에서 화제가 됨.
- 텍스트 생성, 이미지 생성 (DALL·E 연동), 음성 대화, 이미지 이해 등 멀티모달 기반 AI

## ChatGPT에게 질문하기

- 프롬프트 (Prompt)
  - AI에게 내리는 “지시문(command)” 혹은 “설계도(script)”
  - “무엇을, 어떤 방식으로, 어떤 분위기로” 표현할지 명확히 전달
  - 생성형 AI에서 결과의 80%는 프롬프트 품질에 의해 결정됨
  - 입력된 문장은 텍스트 인코더(Text Encoder, 예: CLIP)에 의해 벡터화됨
  - 프롬프트 = 언어 기반 시각 명령어(Language-driven Visual Instruction)

- 프롬프트 엔지니어링(Prompt Engineering)
  - AI가 이해하기 가장 좋은 형태로 명령어를 설계하는 기술
  - 원하는 결과를 반복 가능하게 재현하는 것 (재현성 Reproducibility)

 <div align="center"><img src="https://haesunbyun.github.io/common/images/img_en28.png" height="300"></div>

  - Input prompt (positive) + Negative prompt (요소 제거)

  - 프롬프트 엔지니어는 AI 아티스트의 “디렉터” 역할 수행
  - 현대 디자인 실무에서도 새로운 전문 직군으로 등장
  - ChatGPT, Midjourney, DALL·E, Stable Diffusion 등 공통적으로 사용되는 핵심 기술

- 효과적인 AI 프롬프트 원칙
  - 명확하고 구체적인 요구 사항 제시
  - 역할 부여하기
  - 형식 지정하기
  - AI의 이해를 돕는 맥락 제공하기
  - 스타일 요청하기  
  - 적절한 제약 조건과 지침 제공
  - 이어지는 질문(요청)하기
  - 검증하는 질문 하기

# 3.생성형 AI: 이미지

- 생성형 AI → 멀티모달(Multimodal) 융합 가속화
  - 텍스트 + 이미지 + 음성 + 영상 통합 모델 등장

  - 예전에는 AI가 한 가지 데이터만 처리하는 경우가 많았음
    - 텍스트만 이해 (예: 글쓰기 AI)
    - 이미지 생성만 가능 (예: 그림 생성 AI)

  - 지금은 여러 종류의 데이터를 동시에 다루는 방향으로 빠르게 발전

대표 이미지 생성 모델
- DALL·E 3 (OpenAI)
  - 텍스트 이해력 우수, 자연스러운 합성 표현
  - ChatGPT 인터페이스와 통합 가능
  - 말로 설명하면 그림으로 잘 풀어주는 AI
- Midjourney (Independent Lab)
  - 예술적·디자인 중심, 조명·질감 표현에 강함
  - 디자인·아트 특화 AI
- Stable Diffusion (Open Source)
  - 공개모델, 로컬 실행 가능, 커스터마이징 용이
  - 다양한 파생모델 존재 (e.g., SDXL, DreamShaper)
  - 개발자/고급 사용자용 이미지 AI
- Leonardo AI / Canva AI
  - 디자인 편집과 통합, 사용자 친화적 GUI 제공
  - 디자인 도구 + AI 기능 통합 플랫폼

국내 이미지 생성 기술
- NAVER HyperCLOVA X - SmartCanvas
  - 한국어 프롬프트 인식률 향상
  - 한글 텍스트 포함 이미지 생성 가능

## Gemini

- Gemini의 소개
  - **Gemini**는 Google에서 개발한 멀티모달 생성형 인공지능 모델
  - 텍스트뿐만 아니라 이미지, 코드, 문서, 음성
등 다양한 형태의 정보를 이해하고 생성할 수 있음
  - 즉, 여러 종류의 데이터를 동시에 이해하고 처리할 수 있는 생성형 AI

- 멀티모달 AI
  - 다음과 같은 입력을 동시에 이해할 수 있다.
  - 텍스트, 이미지, 코드, 문서,
  - 예, 이미지를 올리고 질문 가능
→ "이 그래프가 의미하는 것을 설명해줘"

- Google 서비스와 연동
  - Gemini는 Google 서비스와 연동되어 활용된다.

이미지 생성에서 프롬프트

- 구성요소
  - 주제(Subject): 무엇을 그릴 것인가
  - 스타일(Style): 시각적 형태(사진, 수채화, 애니메이션 등)
  - 배경/구도(BG & Composition): 인물/사물의 위치나 시점
  - 조명/색감(Lighting & Color): 톤, 명암, 분위기 결정
  - 감정/상황(Mood & Context): 감성적 메시지 추가

- 텍스트 길이는 100~200자 이내가 적당 (너무 길면 모델 해석 불안정)

- 좋은 프롬프트의 특징
  - 명확성(Clarity): 구체적 명사, 형용사, 동작어 사용 (“sunset sky” vs “sky”)
  - 일관성(Consistency): 모순된 요구 제거 (“dark yet bright tone” → 혼란 유발)
  - 맥락(Context): 장소, 시대, 분위기를 함께 제시 (“Seoul street in 2090”)
  - 목적성(Purpose): “poster”, “profile photo”, “concept art”
  - 제약(Constraint): 크기, 구도, 색상, 화질 등 제한 조건 지정

# 5.생성형 AI: 문서/콘텐츠


- 문서란 무엇인가?
  - 문서는 정보를 기록하고 전달하기 위한 구조화된 텍스트와 시각적 자료의 모음
  - 보고서·기획서·제안서·회의록·슬라이드

  - 말로만 전달할 수 없는 내용들을 정확하고 일관되게 보관하는 역할을 수행
  - 조직·학교·팀 프로젝트 등 협업 환경에서는 문서로 의사소통
  - 문제를 설명하고, 해결 방안을 제시하고, 합의를 이끌어내는 핵심 도구

- 문서의 핵심 기능 (3가지):
기록(Record), 공유(Share), 설득(Persuade)

 <font color="red">Q</font>. 관심이 있는 주제의 최신 뉴스를 찾아 요약을 요청해 봅시다.

- 예시 Prompt 1: 다음 내용을 요약해줘

- 예시 prompt 2: 다음 내용을 3문장으로 요약해줘

- 예시 prompt 3: 다음 내용을 요약하되 용어들은 검색해서 추가해줘

- 예시 prompt4:	다음 내용을 구조화 하고 간단히 요약해줘

- 생성형 AI 기반 보고서 생성의 원리
  - 생성형 AI는 입력된 텍스트를 기반으로 요약, 구조화, 재구성, 확장을 수행
  - AI는 GPT 구조를 이용해 작성 목적에 따라 정보를 재배열하고 문서 구조를 자동 생성

  - 사용자는 AI에게 문서 목적·대상·톤·구조를 명확히 지시하면 높은 품질의 보고서를 획득
  - 생성형 AI는 문장 자연스러움, 논리적 연결, 형식 통일, 톤 조절 등 언어적 작업을 빠르게 수행

  - AI는 “초안 생성기”로 활용하고, 사람은 “검토·수정·근거 검증”에 집중

- 문서 자동화의 가치
  - 문서는 정보 정리와 구조화가 핵심 작업이지만 시간이 많이 들며 반복적
  - 생성형 AI는 요약·확장·재구성에 탁월한 성능
  - PDF, 기사, 논문, 회의록 등을 빠르게 정리가 가능

- 4단계 과정

① 요약(Summarize)

② 구조화(Outline)

③ 확장(Expand)

④ 형식 변환(Transform Format)

## NotebookLM

- NotebookLM 소개
  - 사용자가 업로드한 자료를 기반으로 내용을 분석하고 질문에 답하거나 요약을 생성해 주는 AI 연구 보조 도구
  - 일반적인 생성형 AI가 인터넷 전체 정보를 기반으로 답변을 생성하는 것과 달리, NotebookLM은 사용자가 제공한 문서 자료를 중심으로 답변을 생성함.

- NotebookLM의 주요 특징
  - 자료 기반 AI (Source-grounded AI)
    - 사용자가 업로드한 문서를 기반으로 답변 생성
    - 답변의 출처를 함께 제시
  - 문서 분석 및 요약
    - 업로드된 자료를 분석하여 핵심 내용 요약,
주요 개념 정리, 질문 답변 생성 등을 수행  
  - 연구 및 학습 지원
    - NotebookLM은 연구 및 학습을 지원하는 AI 도구로 설계(논문 정리, 보고서 작성 보조, 강의 자료 분석, 아이디어 정리)
